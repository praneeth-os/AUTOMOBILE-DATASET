```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/Automobile_data.csv')
display(df)
```

```python
print("First 5 Records")
display(df.head())

print("\nDataset Shape")
display(df.shape)

print("\nColumn Names")
display(df.columns)

print("\nData Types")
display(df.dtypes)

print("\nMissing Values")
display(df.isnull().sum())

print("\nStatistical Summary")
display(df.describe())
```

```python
sns.countplot(x='num-of-doors',data=df)
plt.title('Number of Doors')
plt.show()
```

```python
sns.countplot(x='num-of-cylinders',data=df)
plt.title('Number of Cylinders')
plt.show()
```

```python
sns.countplot(x='symboling',data=df)
plt.title('symboling')
plt.show()
```

```python
sns.countplot(x='fuel-type',data=df)
plt.title('fuel-type')
plt.show()
```

```python
sns.countplot(x='drive-wheels',data=df)
plt.title('drive-wheels')
plt.show()
```

```python
sns.countplot(x='body-style',data=df)
plt.title('body-style')
plt.show()
```

```python
plt.figure(figsize=(20,10))
sns.scatterplot(data=df,x='stroke',y='compression-ratio')
plt.title('Stroke vs Compression-Ratio')
plt.show()
```

```python
plt.figure(figsize=(16,10))
sns.scatterplot(data=df,x='city-mpg',y='highway-mpg')
plt.title('Engine Size vs Price')
plt.show()
```

```python
plt.figure(figsize=(20,10))
sns.scatterplot(data=df,x='length',y='curb-weight',hue = 'wheel-base',palette = 'viridis')
plt.title('Length vs Curb Weight')
plt.show()
```

```python
plt.figure(figsize=(20,6))
sns.scatterplot(data=df,x='length',y='curb-weight')
plt.title('Engine Size vs Price')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```
